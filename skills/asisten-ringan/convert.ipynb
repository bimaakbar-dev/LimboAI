# Konversi adapter skill-asisten-ringan → GGUF

Upload `skill-asisten-ringan.zip` (hasil training Colab), lalu notebook ini menghasilkan
`skill-asisten-ringan.gguf` untuk dipakai `llama-server --lora`.

````python
from google.colab import files
import zipfile, os

print("Upload skill-asisten-ringan.zip:")
uploaded = files.upload()
name = next(iter(uploaded))

with zipfile.ZipFile(name) as archive:
    archive.extractall("/content")

adapter = "/content/skill-asisten-ringan"
print("Isi adapter:", sorted(os.listdir(adapter)))
````
