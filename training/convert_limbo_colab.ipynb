# Konversi adapter limbo-lora → GGUF

Upload `limbo-lora.zip` (hasil training Colab), lalu notebook ini menghasilkan
`limbo-lora.gguf` untuk dipakai `llama-server --lora`.

Tidak perlu GPU. Runtime CPU cukup.

````python
from google.colab import files
import zipfile, os

print("Upload limbo-lora.zip:")
uploaded = files.upload()
name = next(iter(uploaded))

with zipfile.ZipFile(name) as archive:
    archive.extractall("/content")

adapter = "/content/limbo-lora"
print("Isi adapter:", sorted(os.listdir(adapter)))
````
