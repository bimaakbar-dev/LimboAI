```python
import json, torch
from pathlib import Path
from torch.utils.data import Dataset
from transformers import (
    AutoTokenizer, AutoModelForCausalLM,
    Trainer, TrainingArguments, DataCollatorForSeq2Seq,
)
from peft import LoraConfig, get_peft_model

BASE_MODEL = "Qwen/Qwen2.5-3B-Instruct"
MAX_LEN = 640

tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

records = [
    json.loads(line)
    for line in Path("/content/dataset.jsonl").read_text(encoding="utf-8").splitlines()
    if line.strip()
]
print("Jumlah contoh:", len(records))

class SFTData(Dataset):
    def __init__(self, rows, tok, max_len):
        self.examples = []
        for row in rows:
            messages = row["messages"]
            prompt = tok.apply_chat_template(messages[:-1], tokenize=False, add_generation_prompt=True)
            answer = messages[-1]["content"] + tok.eos_token
            p_ids = tok(prompt, add_special_tokens=False)["input_ids"]
            a_ids = tok(answer, add_special_tokens=False)["input_ids"]
            ids = (p_ids + a_ids)[:max_len]
            labels = ([-100] * len(p_ids) + a_ids)[:max_len]
            if len(ids) >= 16:
                self.examples.append({
                    "input_ids": ids,
                    "attention_mask": [1] * len(ids),
                    "labels": labels,
                })
    def __len__(self):
        return len(self.examples)
    def __getitem__(self, index):
        return self.examples[index]

train_data = SFTData(records, tokenizer, MAX_LEN)
print("Contoh ter-tokenisasi:", len(train_data))
```

```python
model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    device_map={"": 0},
    torch_dtype=torch.float16,
)
model.config.use_cache = False

lora = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
)
model = get_peft_model(model, lora)
model.print_trainable_parameters()
```

```python
from transformers import TrainingArguments

# Argumen seminimal mungkin: kompatibel dengan transformers versi terbaru.
args = TrainingArguments(
    output_dir="/content/asisten-out",
    num_train_epochs=2,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    learning_rate=2e-4,
    logging_steps=20,
    save_strategy="no",
    report_to="none",
    seed=42,
)

trainer = Trainer(
    model=model,
    args=args,
    train_dataset=train_data,
    data_collator=DataCollatorForSeq2Seq(tokenizer, padding=True, label_pad_token_id=-100),
)
trainer.train()
```

```python
import torch

OUT = "/content/finetune-router"
model.save_pretrained(OUT)
tokenizer.save_pretrained(OUT)
print("Adapter tersimpan:", OUT)

system = "Kamu adalah router otak AI Limbo. Balas HANYA JSON valid dengan kunci intent, route, reason, escalation_prompt."
prompt = tokenizer.apply_chat_template(
    [
        {"role": "system", "content": system},
        {"role": "user", "content": "cek spec pc"},
    ],
    tokenize=False,
    add_generation_prompt=True,
)
inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
with torch.no_grad():
    output = model.generate(
        **inputs, max_new_tokens=120, do_sample=False,
        pad_token_id=tokenizer.eos_token_id,
    )
print(tokenizer.decode(output[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True))
```
